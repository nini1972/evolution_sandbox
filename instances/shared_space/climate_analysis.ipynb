# NYC Climate Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the climate data
df = pd.read_csv('../../shared_space/climate_data.csv')
df['date'] = pd.to_datetime(df['date'])

# Exploratory data analysis
print(df.head())
print(df.info())

# Temperature analysis
plt.figure(figsize=(12, 6))
sns.lineplot(x='date', y='tmax', data=df, label='Maximum Temperature')
sns.lineplot(x='date', y='tmin', data=df, label='Minimum Temperature')
plt.title('Temperature Trends')
plt.xlabel('Date')
plt.ylabel('Temperature (°F)')
plt.legend()
plt.savefig('../../shared_space/temperature_plot.png')

# Precipitation analysis
plt.figure(figsize=(12, 6))
sns.barplot(x='date', y='prcp', data=df)
plt.title('Precipitation Levels')
plt.xlabel('Date')
plt.ylabel('Precipitation (inches)')
plt.savefig('../../shared_space/precipitation_plot.png')

# Wind analysis
plt.figure(figsize=(12, 6))
sns.lineplot(x='date', y='awnd', data=df)
plt.title('Average Wind Speed')
plt.xlabel('Date')
plt.ylabel('Wind Speed (mph)')
plt.savefig('../../shared_space/wind_plot.png')
```
